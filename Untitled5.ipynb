```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image_paths = [
    '/content/Picture2.jpg',
    '/content/Picture3.jpg',
    '/content/Picture4.jpg',
    '/content/Picture5.jpg',
    '/content/Picture6.jpg'
]

images = []

for path in image_paths:
    image = cv2.imread(path, cv2.IMREAD_GRAYSCALE)

    if image is None:
        print("Image not found:", path)
        continue

    images.append(image)

if len(images) == 0:
    print("No images were loaded.")
    exit()

def show_images(threshold_val, gamma_val):

    plt.figure(figsize=(20, 16))

    for i in range(len(images)):

        src_image = images[i]

        negative_image = 255 - src_image

        _, thresholded_image = cv2.threshold(
            src_image,
            threshold_val,
            255,
            cv2.THRESH_BINARY
        )

        table = np.array([
            ((j / 255.0) ** gamma_val) * 255
            for j in range(256)
        ]).astype("uint8")

        gamma_image = cv2.LUT(src_image, table)

        plt.subplot(4, len(images), i + 1)
        plt.imshow(src_image, cmap='gray', vmin=0, vmax=255)
        plt.title(f"Original {i + 1}")
        plt.axis('off')

        plt.subplot(4, len(images), len(images) + i + 1)
        plt.imshow(negative_image, cmap='gray', vmin=0, vmax=255)
        plt.title(f"Negative {i + 1}")
        plt.axis('off')

        plt.subplot(4, len(images), 2 * len(images) + i + 1)
        plt.imshow(thresholded_image, cmap='gray', vmin=0, vmax=255)
        plt.title(f"Threshold {i + 1}")
        plt.axis('off')

        plt.subplot(4, len(images), 3 * len(images) + i + 1)
        plt.imshow(gamma_image, cmap='gray', vmin=0, vmax=255)
        plt.title(f"Gamma {i + 1}")
        plt.axis('off')

    plt.suptitle(
        f"Gray-Level Transformations | Threshold = {threshold_val} | Gamma = {gamma_val}",
        fontsize=16,
        fontweight='bold'
    )

    plt.tight_layout()
    plt.show()

values = [
    (50, 2.0),
    (75, 1.5),
    (100, 1.2),
    (125, 1.0),
    (150, 0.8),
    (175, 0.5),
    (200, 0.3)

]

for threshold_val, gamma_val in values:

    print(
        f"Threshold = {threshold_val}, "
        f"Gamma = {gamma_val}"
    )

    show_images(
        threshold_val,
        gamma_val
    )
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("/content/Avani.jpg")

kernel = np.ones((3, 3), np.float32) / 9

smooth = cv2.filter2D(img, -1, kernel)

plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(smooth, cv2.COLOR_BGR2RGB))
plt.title("Smoothed Image")
plt.axis("off")

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("/content/Avani.jpg")

kernel = np.array([
    [ 0, -1,  0],
    [-1,  5, -1],
    [ 0, -1,  0]
])

sharp = cv2.filter2D(img, -1, kernel)

plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(sharp, cv2.COLOR_BGR2RGB))
plt.title("Sharpened Image")
plt.axis("off")

plt.show()
```
